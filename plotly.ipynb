```python
import pandas as pd
import plotly.express as px

# Sample dataset
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "Jan", "Feb", "Mar", "Apr"],
    "sales": [200, 220, 250, 270, 150, 180, 200, 210],
    "product": ["Laptop", "Laptop", "Laptop", "Laptop",
                "Mobile", "Mobile", "Mobile", "Mobile"]
}

df = pd.DataFrame(data)
fig = px.line(
    df,
    x="month",
    y="sales",
    color="product",          # multiple lines
    markers=True,             # show markers
    line_dash="product",      # different line styles
    title="Monthly Sales Comparison"
)

fig.update_layout(
    title="📊 Advanced Sales Trend",
    xaxis_title="Month",
    yaxis_title="Sales",
    template="plotly_dark"    # theme
)


fig.update_traces(
    marker=dict(size=10),
    line=dict(width=3)
)

fig.show()
```
